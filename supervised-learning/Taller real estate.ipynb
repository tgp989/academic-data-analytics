```python
from sklearn.model_selection import train_test_split

X = df.drop('Y house price of unit area', axis=1)
y = df['Y house price of unit area']

X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.3, random_state=42)
```

```python
X_train = X_train.drop('X6 longitude', axis=1)
X_test = X_test.drop('X6 longitude', axis=1)
```

```python
import statsmodels.api as sm

# Intercepto
X_const = sm.add_constant(X_train)

# Modelo de regresión
modelo_sm = sm.OLS(y_train, X_const).fit()

# Imprimir el resumen del modelo (incluye la ecuación y ANOVA)
print(modelo_sm.summary())

# Obtener y mostrar los residuales
residuales = modelo_sm.resid
print("\nResiduales:")
print(residuales)
```

```python
def backward_elimination(X_train, y_train):

    selected_features = list(X_train.columns)
    X_temp_initial = sm.add_constant(X_train[selected_features])
    model_initial = sm.OLS(y_train, X_temp_initial).fit()
    best_aic = model_initial.aic
    print(f"AIC inicial with all features: {best_aic:.2f}")

    while selected_features:
        feature_to_remove = None
        improved = False
        worst_aic = best_aic # Inicializar con el mejor AIC

        for feature in selected_features:
            temp_features = [f for f in selected_features if f != feature]
            if not temp_features:
                 X_temp = pd.DataFrame({'const': np.ones(len(X_train))})
            else:
                X_temp = sm.add_constant(X_train[temp_features])

            model = sm.OLS(y_train, X_temp).fit()
            current_aic = model.aic

            # En backward elimination, se eliminan las variables que no están, cuando se obtuvo el mejor AIC
            if current_aic < best_aic:
                best_aic = current_aic
                feature_to_remove = feature
                improved = True

        if improved:
            selected_features.remove(feature_to_remove)
            print(f"Variables removidas {feature_to_remove}. Nuevo mejor AIC: {best_aic:.2f}")
        else:
            break

    return selected_features

selected_features_backward = backward_elimination(X_train, y_train)
print("\nVariables seleccionadas usando backward elimination:", selected_features_backward)
```

```python
from sklearn.metrics import mean_squared_error, mean_absolute_error
import numpy as np

# Calculate predictions
X_test_const = sm.add_constant(X_test[selected_features_backward])
predictions = modelo_sm.predict(X_test_const)

# Calculate metrics
r2 = modelo_sm.rsquared
adj_r2 = modelo_sm.rsquared_adj
rmse = np.sqrt(mean_squared_error(y_test, predictions))
mae = mean_absolute_error(y_test, predictions)
# Avoid division by zero in MAPE
mape = np.mean(np.abs((y_test - predictions) / y_test)) * 100

print(f"R2: {r2:.2f}")
print(f"Adjusted R2: {adj_r2:.2f}")
print(f"RMSE: {rmse:.2f}")
print(f"MAE: {mae:.2f}")
print(f"MAPE: {mape:.2f}%")
```

```python
current_aic = modelo_sm.aic
current_bic = modelo_sm.bic
print("aic:", current_aic)
print("bic:", current_bic)
```

Eliminé la variable X6 longitude, porque tiene una correlación importante a la variable X5 Latitude, además porque al comparar ambas con Y, Latitude tiene una correlación mayor.

Dejé la variable Transaction Date, porque aunque su correlación lineal es baja, el precio de las viviendas suele estar influenciado por el momento de la transacción (ciclos económicos, inflación, burbujas inmobiliarias).

## Regresión regularizada

```python
from sklearn.linear_model import Ridge, Lasso, ElasticNet
from sklearn.metrics import mean_squared_error, mean_absolute_error, r2_score
import numpy as np

# Define the features (X) and target (y)
# Using all features from X_train and X_test as SFS is not defined
X_train_regularized = X_train
X_test_regularized = X_test

# Add a small epsilon to prevent division by zero in MAPE
epsilon = 1e-10

# Ridge Regression
ridge_model = Ridge(alpha=1.0) # You can tune alpha
ridge_model.fit(X_train_regularized, y_train)
y_pred_ridge = ridge_model.predict(X_test_regularized)

print("--- Ridge Regression ---")
print(f"RMSE: {np.sqrt(mean_squared_error(y_test, y_pred_ridge)):.2f}")
print(f"MAE: {mean_absolute_error(y_test, y_pred_ridge):.2f}")
mape_ridge = np.mean(np.abs((y_test - y_pred_ridge) / (y_test + epsilon))) * 100
print(f"MAPE: {mape_ridge:.2f}%")
print(f"R2: {r2_score(y_test, y_pred_ridge):.2f}")

# Lasso Regression
lasso_model = Lasso(alpha=1.0) # You can tune alpha
lasso_model.fit(X_train_regularized, y_train)
y_pred_lasso = lasso_model.predict(X_test_regularized)

print("\n--- Lasso Regression ---")
print(f"RMSE: {np.sqrt(mean_squared_error(y_test, y_pred_lasso)):.2f}")
print(f"MAE: {mean_absolute_error(y_test, y_pred_lasso):.2f}")
mape_lasso = np.mean(np.abs((y_test - y_pred_lasso) / (y_test + epsilon))) * 100
print(f"MAPE: {mape_lasso:.2f}%")
print(f"R2: {r2_score(y_test, y_pred_lasso):.2f}")

# Elastic Net Regression
elastic_net_model = ElasticNet(alpha=1.0, l1_ratio=0.5) # You can tune alpha and l1_ratio
elastic_net_model.fit(X_train_regularized, y_train)
y_pred_elastic_net = elastic_net_model.predict(X_test_regularized)

print("\n--- Elastic Net Regression ---")
print(f"RMSE: {np.sqrt(mean_squared_error(y_test, y_pred_elastic_net)):.2f}")
print(f"MAE: {mean_absolute_error(y_test, y_pred_elastic_net):.2f}")
mape_elastic_net = np.mean(np.abs((y_test - y_pred_elastic_net) / (y_test + epsilon))) * 100
print(f"MAPE: {mape_elastic_net:.2f}%")
print(f"R2: {r2_score(y_test, y_pred_elastic_net):.2f}")

print("--- Linear Regression Model Metrics (Test Set) ---")
print(f"RMSE: {rmse:.2f}")
print(f"MAE: {mae:.2f}")
print(f"R2: {r2:.2f}")
print(f"Adjusted R2: {adj_r2:.2f}")
print(f"MAPE: {mape:.2f}%")
```

```python
import numpy as np
from sklearn.metrics import mean_squared_error

# Function to calculate approximate AIC and BIC for scikit-learn models
def calculate_aic_bic(n, mse, k):
    """
    Calculates approximate AIC and BIC.

    Args:
        n (int): Number of observations.
        mse (float): Mean squared error of the model.
        k (int): Number of parameters (including intercept).

    Returns:
        tuple: (AIC, BIC)
    """
    # Approximate log-likelihood based on MSE
    log_likelihood = -n / 2 * np.log(2 * np.pi * mse) - n / 2

    # AIC = -2 * log_likelihood + 2 * k
    aic = -2 * log_likelihood + 2 * k

    # BIC = -2 * log_likelihood + k * log(n)
    bic = -2 * log_likelihood + k * np.log(n)

    return aic, bic

n_train = X_train_regularized.shape[0]
k_ridge = X_train_regularized.shape[1] + 1 # +1 for intercept
k_lasso = X_train_regularized.shape[1] + 1 # +1 for intercept
k_elastic_net = X_train_regularized.shape[1] + 1 # +1 for intercept


# Calculate MSE on the training data for each model
y_pred_train_ridge = ridge_model.predict(X_train_regularized)
mse_train_ridge = mean_squared_error(y_train, y_pred_train_ridge)

y_pred_train_lasso = lasso_model.predict(X_train_regularized)
mse_train_lasso = mean_squared_error(y_train, y_pred_train_lasso)

y_pred_train_elastic_net = elastic_net_model.predict(X_train_regularized)
mse_train_elastic_net = mean_squared_error(y_train, y_pred_train_elastic_net)


# Calculate AIC and BIC for each model
aic_ridge, bic_ridge = calculate_aic_bic(n_train, mse_train_ridge, k_ridge)
aic_lasso, bic_lasso = calculate_aic_bic(n_train, mse_train_lasso, k_lasso)
aic_elastic_net, bic_elastic_net = calculate_aic_bic(n_train, mse_train_elastic_net, k_elastic_net)

print("--- Approximate AIC and BIC ---")
print(f"Linear Regression: AIC={current_aic:.2f}, BIC={current_bic:.2f}")
print(f"Ridge: AIC={aic_ridge:.2f}, BIC={bic_ridge:.2f}")
print(f"Lasso: AIC={aic_lasso:.2f}, BIC={bic_lasso:.2f}")
print(f"Elastic Net: AIC={aic_elastic_net:.2f}, BIC={bic_elastic_net:.2f}")
```

# 📊 Informe Ejecutivo y Técnico – Predicción de Precios de Vivienda

## 1. Objetivo del análisis
El propósito del estudio fue comparar distintos enfoques de regresión estadística y de machine learning para **predecir el precio de vivienda por unidad de área**. Además, se buscó identificar las variables más determinantes en el valor de los inmuebles y traducir los hallazgos en **conclusiones de negocio** que apoyen decisiones estratégicas en el sector inmobiliario.

---

## 2. Metodología

1. **Dataset**:  
   - 414 registros de transacciones inmobiliarias.  
   - Variables:  
     - `X1 transaction date`: fecha de transacción (indicador temporal).  
     - `X2 house age`: antigüedad de la vivienda.  
     - `X3 distance to the nearest MRT station`: distancia a la estación de metro más cercana.  
     - `X4 number of convenience stores`: número de tiendas cercanas.  
     - `X5 latitude` y `X6 longitude`: coordenadas geográficas.  
     - `Y house price of unit area`: precio por metro cuadrado (variable objetivo).

2. **Modelos comparados**:  
   - Regresión Lineal (baseline).  
   - Ridge Regression (regularización L2).  
   - Lasso Regression (regularización L1, selección de variables).  
   - Elastic Net (combinación de L1 y L2).

3. **Criterios de evaluación**:  
   - Métricas de error: RMSE, MAE, MAPE.  
   - Bondad de ajuste: R² y R² ajustado.  
   - Criterios de información: AIC y BIC (parcimonia del modelo).

## 4. Análisis de coeficientes

### 4.1 Regresión Lineal (OLS)
- `X4 number of convenience stores`: coeficiente positivo y altamente significativo (**+2.67, p<0.01**). Cada tienda adicional cercana aumenta en promedio el valor del m².  
- `X1 transaction date`: coeficiente positivo (**+4.77, p=0.044**), indicando **tendencia al alza de precios en el tiempo**.  
- `X2 house age`: coeficiente negativo (-0.24), lo que refleja la pérdida de valor con la antigüedad.  
- `X3 distance to MRT`: coeficiente pequeño y no significativo, sugiriendo poco impacto en este contexto.  
- `X5 latitude` y `X6 longitude`: significativos en el modelo lineal, pero eliminados por Lasso → su efecto es débil y posiblemente redundante.

### 4.2 Regularización
- **Lasso** anuló las variables `transaction date` y `latitude`, confirmando que su efecto no es robusto en presencia de otras variables.  
- **Ridge** y **Elastic Net** redujeron coeficientes pero no mejoraron la precisión.

---

## 5. Implicaciones de negocio

1. **Cercanía a servicios comerciales**  
   - Es el factor más relevante para la valorización de las viviendas.  
   - Estrategia: priorizar desarrollos en zonas con alta densidad de tiendas y servicios básicos.  

2. **Tendencia temporal positiva**  
   - El mercado inmobiliario presenta una clara **apreciación a lo largo del tiempo**.  
   - Estrategia: proyectos adquiridos y vendidos en años recientes capturan mejor valorización.  

3. **Edad de la vivienda como depreciador**  
   - La antigüedad reduce el precio, aunque de manera menos marcada que los servicios.  
   - Estrategia: programas de remodelación o reposición pueden contrarrestar esta pérdida de valor.  

4. **Transporte público**  
   - La distancia al metro no resulta determinante en este dataset.  
   - Implica que en este mercado específico los compradores valoran más la conveniencia cotidiana (tiendas, servicios) que la conectividad.  

5. **Ubicación geográfica (lat/long)**  
   - Importan en el modelo simple, pero no son robustas frente a regularización.  
   - Esto indica que **la ubicación por sí sola no explica tanto como la cercanía a servicios específicos**.

---

## 6. Conclusiones

- **El modelo lineal simple es el más adecuado**, explicando alrededor del **59% de la variación en precios**.  
- **El principal motor de valorización es la densidad de tiendas cercanas**, seguido por el factor tiempo (mercado al alza).  
- **La edad de la vivienda impacta negativamente**, pero en menor medida.  
- **El transporte público no es un factor decisivo en este caso**, lo que orienta a estrategias basadas más en servicios comerciales que en movilidad.  
- Para futuras predicciones, incluir variables macroeconómicas (PIB, tasas de interés, inflación) podría mejorar la capacidad predictiva.
